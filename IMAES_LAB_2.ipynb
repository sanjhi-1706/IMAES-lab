```python
import time


#-----------KNOWLEDGE BASE--------------
# ---------------------------------------------------
# RULE BASE
# Each rule has:
# - id
# - conditions
# - conclusion
# ---------------------------------------------------

rules = [
    {
        "id": "R1",
        "conditions": ["fever", "cough"],
        "conclusion": "flu"
    },
    {
        "id": "R2",
        "conditions": ["fever", "rash"],
        "conclusion": "measles"
    },
    {
        "id": "R3",
        "conditions": ["flu", "body_pain"],
        "conclusion": "needs_rest"
    },
    {
        "id": "R4",
        "conditions": ["cough", "NOT smoker"],
        "conclusion": "respiratory_infection"
    },
    {
        "id": "R5",
        "conditions": ["fever"],
        "conclusion": "visit_doctor"
    }
]


# ---------------------------------------------------
# MATCH FUNCTION
# Checks whether all conditions of a rule are satisfied
# ---------------------------------------------------

def conditions_satisfied(conditions, wm):

    for condition in conditions:

        # Handle negated condition: NOT fact
        if condition.startswith("NOT "):

            fact = condition[4:]

            # Negated condition fails if fact exists in WM
            if fact in wm:
                return False

        else:
            # Normal condition must exist in WM
            if condition not in wm:
                return False

    return True


# ---------------------------------------------------
# FIND CONFLICT SET
# ---------------------------------------------------

def get_conflict_set(rules, wm):

    conflict_set = []

    for rule in rules:

        conditions = rule["conditions"]
        conclusion = rule["conclusion"]

        # Rule is applicable if:
        # 1. All conditions are satisfied
        # 2. Conclusion is not already in WM
        if conditions_satisfied(conditions, wm) and conclusion not in wm:
            conflict_set.append(rule)

    return conflict_set


# ---------------------------------------------------
# DISPLAY FUNCTION
# ---------------------------------------------------

def display_conflict_set(conflict_set):

    if not conflict_set:
        print("Conflict Set: Empty")

    else:
        print("Conflict Set:")

        for rule in conflict_set:
            print(
                f"{rule['id']}: "
                f"IF {' AND '.join(rule['conditions'])} "
                f"THEN {rule['conclusion']}"
            )


# ---------------------------------------------------
# INITIAL WORKING MEMORY
# ---------------------------------------------------

wm = {"fever", "cough"}

print("Working Memory:", wm)

conflict_set = get_conflict_set(rules, wm)

display_conflict_set(conflict_set)
```

```python
import time
import random


# ---------------------------------------------------
# GENERATE RULES
# ---------------------------------------------------

def generate_rules(n):

    generated_rules = []

    for i in range(n):

        rule = {
            "id": f"R{i + 1}",
            "conditions": [
                f"fact_{i % 100}",
                f"fact_{(i + 1) % 100}"
            ],
            "conclusion": f"conclusion_{i}"
        }

        generated_rules.append(rule)

    return generated_rules


# ---------------------------------------------------
# GENERATE WORKING MEMORY
# ---------------------------------------------------

wm = {f"fact_{i}" for i in range(100)}


# ---------------------------------------------------
# TEST DIFFERENT RULE BASE SIZES
# ---------------------------------------------------

sizes = [10, 100, 500, 1000]

for size in sizes:

    generated_rules = generate_rules(size)

    start_time = time.perf_counter()

    conflict_set = get_conflict_set(
        generated_rules,
        wm
    )

    end_time = time.perf_counter()

    elapsed_time = end_time - start_time

    print(
        f"\nNumber of rules: {size}"
    )

    print(
        f"Conflict set size: {len(conflict_set)}"
    )

    print(
        f"Match time: {elapsed_time:.8f} seconds"
    )
```
